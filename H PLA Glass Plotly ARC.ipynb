# H PLA Glass Plotly ARC

```python
import plotly.graph_objects as go
```

```python
categories = ["1% - Before", "2% - Before", "5% - Before", "10% - Before",
             "1% - After", "2% - After", "5% - After", "10% - After"]

modulus_values = [1.686407233, 2.367566095, 1.588429552, 1.574473805,
                  1.9934, 2.0160, 1.4784, 1.4863]

modulus_errors = [0.02487647, 0.048397118, 0.040444719, 0.042685841,
                  0.0097, 0.0197, 0.0067, 0.0081]

color_map = {"Yu": "blueviolet", "JHU": "dodgerblue", "NASA": "limegreen", 
             "WGS": "violet", "Quartz": "violet",
            "Before": "dodgerblue", "After": "limegreen"}

bar_colors = [color_map["Yu"] if "Yu" in cat else
              color_map["Before"] if "Before" in cat else
              color_map["After"] if "After" in cat else
              color_map["JHU"] if "JHU" in cat else
              color_map["NASA"] if "NASA" in cat else
              color_map["WGS"] if "WGS" in cat else
              color_map["Quartz"] for cat in categories]
```

```python
fig = go.Figure()

fig.add_trace(go.Bar(
    x=categories, y=modulus_values,
    marker_color=bar_colors,
    error_y=dict(color='white', array=modulus_errors)))

fig.update_layout(
    paper_bgcolor='black', plot_bgcolor='black',
    font=dict(color='white', size=12),
    title='Hardness for NASA ARC Plasma Tholins on Glass', title_x=0.5,
    xaxis_title='Samples (% CH\N{SUBSCRIPT FOUR})', yaxis_title='Hardness (GPa)',
    margin=dict(t=80, l=80, b=80, r=80), width=900, height=600,
    yaxis_showgrid=True, yaxis_griddash='dot', yaxis_gridcolor='white')

fig.update_xaxes(showline=True, linewidth=2, linecolor='white')
fig.update_yaxes(showline=True, linewidth=2, linecolor='white')

fig.show()

#fig.write_html('/Users/ricardo/UTSA/Research/Yu Lab/Tholin Data Reduction/Plots/Plotly Plots/H_PLA_Glass.html')
```
