```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
# Sample data

categories = ["Yu+2018", 
              "1% JHU", "2% JHU", "5% JHU", "10% JHU",
              "1% NASA", "2% NASA", "5% NASA", "10% NASA",
              "WGS", "Quartz"]

hardness_values = [0.53, 
                   1.049925349, 1.111052122, 1.152779313, 0.844567195,
                   1.9934, 2.0160, 1.4784, 1.4863,
                   1.5, 6.0]

hardness_errors = [0.03, 
                   0.011497663, 0.015034001, 0.008552291, 0.007144113,
                   0.0097, 0.0197, 0.0067, 0.0081,
                   0, 0.5]
```

```python
# Creating color map and assigning bar colors

color_map = {"Yu": "blueviolet", "JHU": "dodgerblue", "NASA": "limegreen", 
             "WGS": "violet", "Quartz": "violet"}

bar_colors = [color_map["Yu"] if "Yu" in cat else 
              color_map["JHU"] if "JHU" in cat else
              color_map["NASA"] if "NASA" in cat else
              color_map["WGS"] if "WGS" in cat else
              color_map["Quartz"] for cat in categories]
```

```python
# Plotting figure

plt.figure(dpi=150, facecolor="white")

plt.bar(np.arange(len(categories)), hardness_values, edgecolor="black", color=bar_colors, yerr=hardness_errors, capsize=5,
        ecolor="black", zorder=2, label="Hardness")
plt.xticks(np.arange(len(categories)), categories, rotation=30, fontsize=9)
plt.yticks(fontsize=9)
plt.xlabel("Samples (% CH$_4$)", color="black")
plt.ylabel("Hardness (GPa)", color="black")
plt.title("Hardness for Plasma Tholins on Glass", color="black")
plt.grid(axis="y", linestyle="dashed", alpha=0.7, color="black")

# When saving fig, save in 1200 dpi ==> plt.savefig('___.png', dpi=1200)

plt.show()
plt.close()
```
